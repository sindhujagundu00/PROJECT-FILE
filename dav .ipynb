```python
#To import Pandas, which is used for reading, checking, cleaning and analyzing datasets.
import pandas as pd
```

```python
#To read the Excel dataset and store it in a DataFrame named df.
df = pd.read_excel("Ecommerce_Customer_Journey_India_2024(2).xlsx",
                   sheet_name="Raw_Dataset")

print(df)
```

```python
#To view the first five records of the dataset.
print(df.head())
```

```python
#To view the last five records of the dataset.
print(df.tail())
```

```python
#To find the total number of rows and columns.
print(df.shape)
```

```python
#To display all column names in the dataset.
print(df.columns)
```

```python
#To identify whether columns contain integer, decimal, text or other data types.
print(df.dtypes)
```

```python
#To check column names, non-null values, data types and memory usage.
df.info()
```

```python
#To count the number of completely duplicated rows in the dataset.
print(df.duplicated().sum())
```

```python
#To display the duplicate records present in the dataset.
print(df[df.duplicated()])
```

```python
#To remove completely identical rows and keep only one copy of each record.
df = df.drop_duplicates()

print(df)
```

```python
#To verify whether duplicate rows have been removed successfully.
print(df.duplicated().sum())
```

```python
#To identify how many empty cells are present in each column.
print(df.isnull().sum())
```

```python
#To calculate the total number of missing cells in the entire dataset.
print(df.isnull().sum().sum())
```

```python
#To remove complete rows that contain one or more missing values.
df = df.dropna()

print(df)
```

```python
#To verify whether missing rows have been removed successfully.
print(df.isnull().sum().sum())
```

```python
#To save the cleaned dataset into a new Excel file without changing the original dataset.
df.to_excel("Cleaned_Ecommerce_Dataset.xlsx", index=False)

print("Dataset saved successfully")
```

```python
#To reload the original dataset whenever you want to perform cleaning again.
df = pd.read_excel("Ecommerce_Customer_Journey_India_2024(2).xlsx",
                   sheet_name="Raw_Dataset")

print(df)
```
